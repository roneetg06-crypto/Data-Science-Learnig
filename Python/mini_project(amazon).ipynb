📊 Revenue by Category → Bar Chart

```python
rev_cat = df.groupby(df['Category'])['TotalAmount'].sum()
plt.bar(rev_cat.index,rev_cat.values)
plt.title("Revenue Salary by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")

plt.show()
```

```python
plt.hist(df["TotalAmount"])

plt.title("Product Amount Distribution")

plt.show()
```

```python
plt.scatter(
    df["Quantity"],
    df["TotalAmount"]
)

plt.xlabel("Quantity")
plt.ylabel("TotalAmount")

plt.show()
```

```python
import seaborn as sns

sns.boxplot(y=df["TotalAmount"])
```

```python
numeric = df.select_dtypes(include="number")

sns.heatmap(
    numeric.corr(),
    annot=True
)
```
