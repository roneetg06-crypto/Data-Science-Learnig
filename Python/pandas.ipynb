```python
import pandas as pd

data = {
    "student_id":[101,102,103,104,105,106,107,108,109,110],
    "name":[
        "Rohan","Priya","Aman","Neha","Rahul",
        "Anjali","Vikas","Sneha","Arjun","Pooja"
    ],
    "gender":[
        "Male","Female","Male","Female","Male",
        "Female","Male","Female","Male","Female"
    ],
    "department":[
        "CSE","CSE","IT","IT","CSE",
        "ECE","ECE","IT","CSE","ECE"
    ],
    "city":[
        "Delhi","Delhi","Mumbai","Mumbai","Pune",
        "Delhi","Pune","Delhi","Mumbai","Pune"
    ],
    "semester":[
        1,1,2,2,3,
        3,1,2,3,1
    ],
    "gpa":[
        8.5,9.2,7.8,8.1,9.0,
        7.5,8.8,8.9,7.9,9.4
    ],
    "python_skill":[
        "Strong","Intermediate","Strong","Beginner","Strong",
        "Intermediate","Strong","Beginner","Intermediate","Strong"
    ],
    "projects":[
        2,3,1,2,4,
        2,5,1,3,4
    ]
}

df = pd.DataFrame(data)

df
```

```python
result = df.pivot_table(
    index='city',
    columns='department',
    values='gpa',
    aggfunc='mean'
)
print(result)
```

```python
def increment(projects):
    result = projects + 1
    return result

df['projects'] = df['projects'].apply(increment)
df
```
